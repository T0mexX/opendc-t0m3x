# Check Workload with ID and DURATION is executed consistently in terms of active flow per instant

```python
import pandas as pd
from pandas import DataFrame as DF
```

```python
class NetWlCol:
    TIMESTAMP = "timestamp"
    TRANSMITTER_ID = "transmitter_id"
    DEST_ID = "dest_id"
    NET_TX = "net_tx"
    FLOW_ID = "flow_id"
    DURATION = "duration"
```

```python
# Some columns are missing.
class NodeOutCol:
    """Many not included here"""
    TS_ABS = "timestamp_absolute"
    MIN_F_TPUT = "min_flow_throughput_ratio"
    MAX_F_TPUT = "max_flow_throughput_ratio"
    AVG_F_YPU_RATIO = "avg_flow_throughput_ratio"
    TOT_TPUT_RATIO = "node_throughput_ratio"
    CURR_PWR_USE = "power_draw_watts"
    EN_USE = "energy_consumed_joule"
```

```python
# Some columns are missing.
class NetOutCol:
    """Many not included here"""
    TS_ABS = "timestamp_absolute"
    NUM_FLOWS = "flows"
    AVG_TPUT_RATIO = "avg_throughput_ratio"
    TOT_TPUT_RATIO = "tot_tput_ratio"
    TOT_TPUT = "tot_tput_mbps"
    WORST_TPUT_RATIO = "worst_tput_ratio"
    CURR_PWR_USE = "power_draw_watt"
    EN_USE = "energy_consumed_joule"
```

```python
import bisect

def get_flows_input() -> list[int]:
    df: DF = pd.read_parquet("../resources/example-traces/anarchy_online_10servers.parquet")
    instants: list[int] = list()
    for inst, dur in zip(df[NetWlCol.TIMESTAMP], df[NetWlCol.DURATION]):
        inst: int = int(inst.value / 1e6)
        dur: int
        if inst not in instants:
            instants.append(inst)
        if (inst + dur) not in instants:
            instants.append(inst + dur)
    
    instants = sorted(instants)


    print(instants)
    def find_range_indices(sorted_list, low, high):
        start_index = bisect.bisect_left(sorted_list, low)
        end_index = bisect.bisect_left(sorted_list, high)
        if start_index >= len(sorted_list) or sorted_list[start_index] > high:
            return None  # No elements in range
        return range(start_index, end_index)

        
    result: list[int] = list(0 for _ in range(len(instants)))
    
    for inst, dur in zip(df[NetWlCol.TIMESTAMP], df[NetWlCol.DURATION]):
        inst: int = int (inst.value / 1e6)
        dur: int
        print(inst, dur, find_range_indices(instants, inst, inst+dur))

        for idx in find_range_indices(instants, inst, inst+dur):
            result[idx] += 1
            
    return result


"""Maps each input trace instant to the number of flows that are active, based on the starting time and duration"""
flows_input: list[int] = get_flows_input()[:-1]
flows_input
```

```python
def get_flows_output() -> list[int]:
    df: DF = pd.read_parquet("../tobedel/networking/network.parquet")
    df: DF = df.sort_values(NetOutCol.TS_ABS)
    
    return df[NetOutCol.NUM_FLOWS].to_list()


"""Maps each input trace instant to the number of flows that are active, based on the starting time and duration"""
flows_output: list[int] = get_flows_output()
flows_output
```
